Metrics Comparison

```python
import matplotlib.pyplot as plt

models = [
    "Baseline",
    "BatchNorm",
    "Dropout",
    "Aug+LR",
    "U-Net 128",
    "U-Net 224"
]

iou = [0.3763, 0.3771, 0.4371, 0.4157, 0.5406, 0.5674]
precision = [0.5423, 0.6369, 0.5728, 0.5731, 0.7612, 0.7401]
recall = [0.5600, 0.4848, 0.6564, 0.6106, 0.6558, 0.7112]
f1 = [0.5429, 0.5435, 0.6058, 0.5841, 0.6980, 0.7199]
mae = [0.2780, 0.2326, 0.2606, 0.2690, 0.1810, 0.1717]

table_data = []

for i in range(len(models)):
    table_data.append([
        models[i],
        f"{iou[i]:.4f}",
        f"{precision[i]:.4f}",
        f"{recall[i]:.4f}",
        f"{f1[i]:.4f}",
        f"{mae[i]:.4f}"
    ])

columns = [
    "Model",
    "IoU",
    "Precision",
    "Recall",
    "F1-score",
    "MAE"
]

fig, ax = plt.subplots(figsize=(12, 3))

ax.axis("off")

table = ax.table(
    cellText=table_data,
    colLabels=columns,
    loc="center",
    cellLoc="center"
)

table.auto_set_font_size(False)
table.set_fontsize(10)

table.scale(1, 2)

plt.title(
    "Metrics Comparison Table",
    fontsize=14,
    fontweight="bold",
    pad=20
)

plt.show()
```

Loss Curve Comparison

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

baseline = mpimg.imread(
    "outputs/baseline/loss_curve_baseline.png"
)

unet = mpimg.imread(
    "outputs/unet/loss_curve_unet.png"
)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].imshow(baseline)
axes[0].set_title("Baseline Loss Curve")
axes[0].axis("off")

axes[1].imshow(unet)
axes[1].set_title("Improved Loss Curve")
axes[1].axis("off")

plt.suptitle(
    "Baseline vs Improved Comparison",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

os.makedirs(
    "outputs/comparison",
    exist_ok=True
)

plt.savefig(
    "outputs/comparison/loss_curve_comparison.png",
    dpi=300
)

plt.show()
```
